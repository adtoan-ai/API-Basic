```python
# ============== CELL 6: HIỂN THỊ MẪU ẢNH THEO TỪNG LỚP ==============
# Chọn ảnh "đẹp": ưu tiên ảnh có độ phân giải lớn và độ tương phản tốt
rng = np.random.RandomState(SEED)
SAMPLE_PATHS = {}

for c in range(NUM_CLASSES):
    sub = train_csv[train_csv['ClassId'] == c].copy()
    sub['area'] = sub['Width'] * sub['Height']
    top = sub.nlargest(40, 'area')                       # 40 ảnh nét nhất
    best, best_score = None, -1
    for fp in top['filepath'].sample(min(15, len(top)), random_state=SEED):
        img = cv2.imread(fp)
        if img is None: continue
        g = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        score = g.std() - abs(g.mean() - 128) * 0.3      # contrast cao + sáng vừa
        if score > best_score: best_score, best = score, fp
    SAMPLE_PATHS[c] = best

fig, axes = plt.subplots(6, 8, figsize=(20, 16))
axes = axes.ravel()
for c in range(NUM_CLASSES):
    img = cv2.cvtColor(cv2.imread(SAMPLE_PATHS[c]), cv2.COLOR_BGR2RGB)
    axes[c].imshow(img); axes[c].axis('off')
    axes[c].set_title(f"[{c}] {CLASS_NAMES[c][:24]}", fontsize=8.5)
for k in range(NUM_CLASSES, len(axes)): axes[k].axis('off')
plt.suptitle('ẢNH MẪU ĐẠI DIỆN CHO TỪNG CLASS (43 classes) — ẢNH GỐC', fontsize=16, y=0.995)
plt.tight_layout(); plt.show()
```

```python
# ============== CELL 7: TRACK-BASED TRAIN/VALIDATION SPLIT ==============
def get_track_id(path):
    """
    Tên file GTSRB: {ClassId}_{TrackId}_{FrameId}.png  (vd 00020_00000_00029.png)
    Mỗi track = 30 ảnh liên tiếp của CÙNG 1 biển báo vật lý.
    → Phải chia theo track, KHÔNG chia theo từng ảnh, nếu không train/val bị trùng
      gần như y hệt nhau ⇒ validation accuracy ảo, che giấu overfitting.
    """
    base = os.path.splitext(os.path.basename(path))[0]
    parts = base.split('_')
    if len(parts) >= 3: return parts[1]
    if len(parts) == 2: return parts[0]
    return base

train_csv['TrackId'] = train_csv['filepath'].apply(get_track_id)
train_csv['TrackKey'] = train_csv['ClassId'].astype(str) + '_' + train_csv['TrackId']

VAL_RATIO = 0.2
val_keys = []
rs = np.random.RandomState(SEED)
for c in range(NUM_CLASSES):
    keys = train_csv.loc[train_csv['ClassId']==c, 'TrackKey'].unique()
    rs.shuffle(keys)
    n_val = max(1, int(round(len(keys)*VAL_RATIO)))
    val_keys.extend(keys[:n_val])
val_keys = set(val_keys)

val_df   = train_csv[train_csv['TrackKey'].isin(val_keys)].reset_index(drop=True)
tr_df    = train_csv[~train_csv['TrackKey'].isin(val_keys)].reset_index(drop=True)

print("="*70); print("KẾT QUẢ CHIA TẬP (THEO TRACK)"); print("="*70)
print(f"Tổng số track           : {train_csv['TrackKey'].nunique():,}")
print(f"Train : {len(tr_df):,} ảnh / {tr_df['TrackKey'].nunique():,} track")
print(f"Valid : {len(val_df):,} ảnh / {val_df['TrackKey'].nunique():,} track")
print(f"Test  : {len(test_csv):,} ảnh")
overlap = set(tr_df['TrackKey']) & set(val_df['TrackKey'])
print(f"✅ Số track bị trùng giữa Train và Val: {len(overlap)}  (phải = 0 → không data leakage)")

# Kiểm tra đủ 43 class ở cả 2 tập
print(f"Số class trong Train: {tr_df['ClassId'].nunique()} | trong Val: {val_df['ClassId'].nunique()}")

plt.figure(figsize=(18,5))
idx = np.arange(NUM_CLASSES); wdt=0.4
plt.bar(idx-wdt/2, tr_df['ClassId'].value_counts().sort_index().reindex(idx, fill_value=0),
        wdt, label='Train', color='steelblue')
plt.bar(idx+wdt/2, val_df['ClassId'].value_counts().sort_index().reindex(idx, fill_value=0),
        wdt, label='Validation', color='orange')
plt.xticks(idx); plt.xlabel('Class ID'); plt.ylabel('Số ảnh')
plt.title('Phân bố class sau khi chia Train / Validation theo track'); plt.legend()
plt.tight_layout(); plt.show()

train_paths, train_labels = tr_df['filepath'].values,  tr_df['ClassId'].values
val_paths,   val_labels   = val_df['filepath'].values, val_df['ClassId'].values
test_paths,  test_labels  = test_csv['filepath'].values, test_csv['ClassId'].values
```

```python
# ====== CELL 8: TIỀN XỬ LÝ RIÊNG BIỆT CHO TỪNG MÔ HÌNH ======
"""
Mỗi mô hình có pipeline riêng, khác nhau ở CẢ 4 KHÂU:
  (1) kích thước đầu vào   (2) phương pháp nội suy
  (3) kỹ thuật tăng cường  (4) cách chuẩn hóa pixel
"""

CFG = {
    'custom': {
        'size'   : SIZE_CUSTOM,
        'interp' : cv2.INTER_AREA,      # thu nhỏ → chống răng cưa
        'enhance': 'clahe',             # CLAHE kênh L (LAB)
        'norm'   : 'rescale',           # [0, 1]
        'desc'   : '48×48 | INTER_AREA | CLAHE(LAB-L) | scale [0,1]',
    },
    'resnet': {
        'size'   : SIZE_RESNET,
        'interp' : cv2.INTER_CUBIC,     # phóng to → giữ độ mượt
        'enhance': 'unsharp',           # làm nét sau khi upscale
        'norm'   : 'caffe',             # RGB→BGR, trừ mean ImageNet
        'desc'   : '128×128 | INTER_CUBIC | Unsharp Mask | caffe (mean-subtract)',
    },
    'mobilenet': {
        'size'   : SIZE_MOBILENET,
        'interp' : cv2.INTER_LINEAR,
        'enhance': 'gamma',             # gamma thích ứng, chi phí thấp
        'norm'   : 'tf',                # [-1, 1]
        'desc'   : '96×96 | INTER_LINEAR | Adaptive Gamma | scale [-1,1]',
    },
}

# ---------------- CÁC HÀM TĂNG CƯỜNG (numpy, ảnh RGB uint8) ----------------
_clahe_small = cv2.createCLAHE(clipLimit=2.5, tileGridSize=(4, 4))

def enh_clahe(img):
    """CLAHE trên kênh L của LAB: tăng tương phản CỤC BỘ, giữ nguyên màu sắc.
       Dùng cho Custom CNN vì model train from scratch, cần tín hiệu rõ nhất."""
    lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
    l, a, b = cv2.split(lab)
    l = _clahe_small.apply(l)
    return cv2.cvtColor(cv2.merge([l, a, b]), cv2.COLOR_LAB2RGB)

def enh_unsharp(img, sigma=1.2, amount=1.3):
    """Unsharp Mask: sharp = img + amount*(img - blur).
       Dùng cho ResNet50 vì ảnh gốc ~40px bị phóng lên 128px → mất nét biên."""
    blur  = cv2.GaussianBlur(img, (0, 0), sigma)
    sharp = cv2.addWeighted(img.astype(np.float32), 1 + amount,
                            blur.astype(np.float32), -amount, 0)
    return np.clip(sharp, 0, 255).astype(np.uint8)

def enh_gamma(img, target=0.5):
    """Adaptive Gamma Correction: đưa độ sáng trung bình về ~0.5.
       gamma = log(target)/log(mean). Chỉ 1 phép LUT → rất nhanh, hợp edge device."""
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    mean = float(np.clip(gray.mean() / 255.0, 0.05, 0.95))
    gamma = float(np.clip(np.log(target) / np.log(mean), 0.4, 2.5))
    lut = np.clip(((np.arange(256) / 255.0) ** gamma) * 255.0, 0, 255).astype(np.uint8)
    return cv2.LUT(img, lut)

ENHANCERS = {'clahe': enh_clahe, 'unsharp': enh_unsharp, 'gamma': enh_gamma, 'none': lambda x: x}

# ---------------- RESIZE + ENHANCE (chạy trên CPU qua numpy_function) ----------------
def make_spatial_fn(model_type):
    """Trả về hàm numpy: ảnh RGB uint8 gốc → ảnh RGB uint8 đã resize + enhance."""
    cfg  = CFG[model_type]
    size, interp = cfg['size'], cfg['interp']
    enh  = ENHANCERS[cfg['enhance']]
    def _fn(img):
        img = cv2.resize(img, (size[1], size[0]), interpolation=interp)
        return enh(img).astype(np.uint8)
    return _fn

# ---------------- CHUẨN HÓA (chạy trên GPU, sau augmentation) ----------------
def preprocess_fn(model_type):
    norm = CFG[model_type]['norm']
    if norm == 'rescale': return lambda x: x / 255.0
    if norm == 'caffe'  : return lambda x: resnet_pp.preprocess_input(x)
    if norm == 'tf'     : return lambda x: mbv2_pp.preprocess_input(x)
    raise ValueError(norm)

# ---------------- AUGMENTATION RIÊNG CHO TỪNG MODEL ----------------
# Custom CNN train from scratch → cần augment MẠNH để tránh overfit.
# Model pretrained đã có prior tốt → augment vừa phải, tránh phá vỡ đặc trưng.
AUGMENTERS = {
    'custom': keras.Sequential([
        layers.RandomTranslation(0.12, 0.12, fill_mode='nearest'),
        layers.RandomZoom(0.18, 0.18, fill_mode='nearest'),
        layers.RandomRotation(0.06, fill_mode='nearest'),
        layers.RandomBrightness(0.30, value_range=(0, 255)),
        layers.RandomContrast(0.25),
    ], name='aug_custom'),

    'resnet': keras.Sequential([
        layers.RandomTranslation(0.08, 0.08, fill_mode='nearest'),
        layers.RandomZoom(0.12, 0.12, fill_mode='nearest'),
        layers.RandomRotation(0.04, fill_mode='nearest'),
        layers.RandomBrightness(0.20, value_range=(0, 255)),
        layers.RandomContrast(0.15),
    ], name='aug_resnet'),

    'mobilenet': keras.Sequential([
        layers.RandomTranslation(0.10, 0.10, fill_mode='nearest'),
        layers.RandomZoom(0.15, 0.15, fill_mode='nearest'),
        layers.RandomRotation(0.05, fill_mode='nearest'),
        layers.RandomBrightness(0.25, value_range=(0, 255)),
        layers.RandomContrast(0.20),
    ], name='aug_mobilenet'),
}
# ❗ KHÔNG dùng RandomFlip cho mọi model: lật ngang làm đổi ý nghĩa biển báo
#    (Turn left ahead ↔ Turn right ahead, Keep left ↔ Keep right).

# ---------------- BUILD DATASET ----------------
def build_dataset(paths, labels, model_type, training=False, sample_w=None, shuffle=False):
    size       = CFG[model_type]['size']
    spatial_fn = make_spatial_fn(model_type)
    norm_fn    = preprocess_fn(model_type)

    def _load(path, label, sw):
        img = tf.io.read_file(path)
        img = tf.io.decode_image(img, channels=3, expand_animations=False)
        img = tf.numpy_function(spatial_fn, [img], tf.uint8)   # resize + enhance riêng
        img.set_shape([size[0], size[1], 3])
        img = tf.cast(img, tf.float32)                         # giữ dải [0,255]
        onehot = tf.one_hot(label, NUM_CLASSES)                # ✅ One-hot encoding
        return img, onehot, sw

    sw = np.ones(len(paths), np.float32) if sample_w is None else sample_w
    ds = tf.data.Dataset.from_tensor_slices((paths, labels.astype(np.int32), sw))
    if shuffle:
        ds = ds.shuffle(min(len(paths), 20000), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(_load, num_parallel_calls=AUTOTUNE).batch(BATCH_SIZE)

    if training:   # LUỒNG TRAIN: resize+enhance → augmentation → normalize
        aug = AUGMENTERS[model_type]
        ds = ds.map(lambda x, y, w: (aug(x, training=True), y, w), AUTOTUNE)
    # LUỒNG VAL/TEST: chỉ resize+enhance → normalize (KHÔNG augmentation)
    ds = ds.map(lambda x, y, w: (norm_fn(x), y, w), AUTOTUNE)
    return ds.prefetch(AUTOTUNE)

# ---------------- Hàm dùng lại cho Grad-CAM / hiển thị ----------------
def prepare_single(path, model_type):
    """Trả về (tensor đã tiền xử lý shape (1,H,W,3), ảnh RGB uint8 để hiển thị)."""
    img  = cv2.cvtColor(cv2.imread(path), cv2.COLOR_BGR2RGB)
    proc = make_spatial_fn(model_type)(img)
    x    = preprocess_fn(model_type)(proc.astype(np.float32))[None, ...]
    return x, proc

print("="*90); print("PIPELINE TIỀN XỬ LÝ RIÊNG CHO TỪNG MÔ HÌNH"); print("="*90)
for mt in ['custom', 'resnet', 'mobilenet']:
    print(f"  {mt:<10} → {CFG[mt]['desc']}")
    print(f"  {'':<10}   Augmentation: {AUGMENTERS[mt].name} "
          f"({len(AUGMENTERS[mt].layers)} phép biến đổi, chỉ áp dụng cho TRAIN)")
print("="*90)
```

```python
# ====== CELL 11: XÂY DỰNG MÔ HÌNH ======
def build_custom_cnn(input_shape=(*SIZE_CUSTOM, 3), n=NUM_CLASSES):
    """CNN thiết kế riêng cho ảnh biển báo nhỏ: chỉ 3 lần downsample
       → feature map cuối 6×6, giữ được chi tiết chữ số/ký hiệu."""
    inp = layers.Input(shape=input_shape, name='input_custom')
    x = inp
    for i, f in enumerate([32, 64, 128]):
        x = layers.Conv2D(f, 3, padding='same', use_bias=False, name=f'b{i+1}_conv1')(x)
        x = layers.BatchNormalization(name=f'b{i+1}_bn1')(x); x = layers.Activation('relu')(x)
        x = layers.Conv2D(f, 3, padding='same', use_bias=False, name=f'b{i+1}_conv2')(x)
        x = layers.BatchNormalization(name=f'b{i+1}_bn2')(x); x = layers.Activation('relu')(x)
        x = layers.MaxPooling2D(2, name=f'b{i+1}_pool')(x)
        x = layers.Dropout(0.20 + 0.05*i, name=f'b{i+1}_drop')(x)
    x = layers.Conv2D(256, 3, padding='same', use_bias=False, name='top_conv')(x)
    x = layers.BatchNormalization(name='top_bn')(x)
    x = layers.Activation('relu', name='last_conv_act')(x)      # ← layer cho Grad-CAM
    x = layers.GlobalAveragePooling2D(name='gap')(x)
    x = layers.Dense(256, activation='relu')(x)
    x = layers.BatchNormalization()(x); x = layers.Dropout(0.5)(x)
    out = layers.Dense(n, activation='softmax', name='predictions')(x)
    return keras.Model(inp, out, name='Custom_CNN')

def build_resnet50(input_shape=(*SIZE_RESNET, 3), n=NUM_CLASSES):
    """ResNet50 transfer learning — input 128×128 để feature map cuối còn 4×4."""
    inp  = layers.Input(shape=input_shape, name='input_resnet')
    base = ResNet50(include_top=False, weights='imagenet', input_tensor=inp)
    base.trainable = False
    x = layers.GlobalAveragePooling2D(name='gap')(base.output)
    x = layers.BatchNormalization()(x); x = layers.Dropout(0.4)(x)
    x = layers.Dense(256, activation='relu')(x); x = layers.Dropout(0.3)(x)
    out = layers.Dense(n, activation='softmax', name='predictions')(x)
    return keras.Model(inp, out, name='ResNet50_TL')

def build_mobilenetv2(input_shape=(*SIZE_MOBILENET, 3), n=NUM_CLASSES):
    """MobileNetV2 transfer learning — input 96×96, size có trọng số ImageNet chính thức,
       hướng tới triển khai trên thiết bị biên."""
    inp  = layers.Input(shape=input_shape, name='input_mbv2')
    base = MobileNetV2(include_top=False, weights='imagenet', input_tensor=inp, alpha=1.0)
    base.trainable = False
    x = layers.GlobalAveragePooling2D(name='gap')(base.output)
    x = layers.BatchNormalization()(x); x = layers.Dropout(0.4)(x)
    x = layers.Dense(256, activation='relu')(x); x = layers.Dropout(0.3)(x)
    out = layers.Dense(n, activation='softmax', name='predictions')(x)
    return keras.Model(inp, out, name='MobileNetV2_TL')

model_custom = build_custom_cnn()
model_resnet = build_resnet50()
model_mbv2   = build_mobilenetv2()

print("="*95)
print(f"{'Model':<18}{'Input':<14}{'Total params':>16}{'Layers':>10}{'Feature map cuối':>20}")
print("="*95)
info = [(model_custom, SIZE_CUSTOM, '6×6'), (model_resnet, SIZE_RESNET, '4×4'),
        (model_mbv2, SIZE_MOBILENET, '3×3')]
for m, s, fm in info:
    print(f"{m.name:<18}{f'{s[0]}×{s[1]}':<14}{m.count_params():>16,}{len(m.layers):>10}{fm:>20}")
print("="*95)

model_custom.summary()
```

```python
# ====== CELL 12: HUẤN LUYỆN 3 MODEL ======
EPOCHS_CUSTOM = 30
EPOCHS_HEAD   = 8       # Phase 1: chỉ train head
EPOCHS_FT     = 20      # Phase 2: fine-tune

def get_callbacks(name, patience=6):
    return [
        keras.callbacks.EarlyStopping(monitor='val_accuracy', patience=patience,
                                      restore_best_weights=True, verbose=1),
        keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.4,
                                          patience=3, min_lr=1e-7, verbose=1),
        keras.callbacks.ModelCheckpoint(f'{OUT_DIR}/{name}_best.keras',
                                        monitor='val_accuracy', save_best_only=True, verbose=0),
    ]

def merge_hist(h1, h2=None):
    h = {k: list(v) for k, v in h1.history.items()}
    if h2:
        for k, v in h2.history.items(): h.setdefault(k, []).extend(list(v))
    return h

def unfreeze_from(model, prefix):
    """Mở khóa toàn bộ layer kể từ layer đầu tiên có tên bắt đầu bằng `prefix`.
       KHÔNG loại trừ BatchNormalization: BN phải được cập nhật lại theo thống kê
       của GTSRB, vì phân bố ảnh biển báo lệch rất xa so với ImageNet."""
    on = False
    for layer in model.layers:
        if layer.name.startswith(prefix): on = True
        layer.trainable = on
    n_tr = sum(1 for l in model.layers if l.trainable)
    print(f"   → Mở khóa {n_tr}/{len(model.layers)} layer (từ '{prefix}' trở đi)")

HIST, TRAIN_TIME = {}, {}

# ---------------- 1) CUSTOM CNN ----------------
print("\n" + "="*80); print("🚀 TRAINING 1/3 — CUSTOM CNN (train from scratch)"); print("="*80)
model_custom.compile(optimizer=keras.optimizers.Adam(1e-3),
                     loss='categorical_crossentropy', metrics=['accuracy'])
t0 = time.time()
h_c = model_custom.fit(DATASETS['custom']['train'],
                       validation_data=DATASETS['custom']['val'],
                       epochs=EPOCHS_CUSTOM, callbacks=get_callbacks('custom', 8), verbose=1)
TRAIN_TIME['Custom CNN'] = time.time() - t0
HIST['Custom CNN'] = merge_hist(h_c)

# ---------------- 2) RESNET50 ----------------
print("\n" + "="*80); print("🚀 TRAINING 2/3 — RESNET50  |  Phase 1: đóng băng base"); print("="*80)
model_resnet.compile(optimizer=keras.optimizers.Adam(1e-3),
                     loss='categorical_crossentropy', metrics=['accuracy'])
t0 = time.time()
h_r1 = model_resnet.fit(DATASETS['resnet']['train'],
                        validation_data=DATASETS['resnet']['val'],
                        epochs=EPOCHS_HEAD, callbacks=get_callbacks('resnet_p1', 5), verbose=1)

print("\n🔧 RESNET50  |  Phase 2: fine-tune từ conv4 (BN được cập nhật lại)")
unfreeze_from(model_resnet, 'conv4')
model_resnet.compile(optimizer=keras.optimizers.Adam(1e-4),
                     loss='categorical_crossentropy', metrics=['accuracy'])
h_r2 = model_resnet.fit(DATASETS['resnet']['train'],
                        validation_data=DATASETS['resnet']['val'],
                        epochs=EPOCHS_FT, callbacks=get_callbacks('resnet_p2', 7), verbose=1)
TRAIN_TIME['ResNet50'] = time.time() - t0
HIST['ResNet50'] = merge_hist(h_r1, h_r2)

# ---------------- 3) MOBILENETV2 ----------------
print("\n" + "="*80); print("🚀 TRAINING 3/3 — MOBILENETV2  |  Phase 1: đóng băng base"); print("="*80)
model_mbv2.compile(optimizer=keras.optimizers.Adam(1e-3),
                   loss='categorical_crossentropy', metrics=['accuracy'])
t0 = time.time()
h_m1 = model_mbv2.fit(DATASETS['mobilenet']['train'],
                      validation_data=DATASETS['mobilenet']['val'],
                      epochs=EPOCHS_HEAD, callbacks=get_callbacks('mbv2_p1', 5), verbose=1)

print("\n🔧 MOBILENETV2  |  Phase 2: fine-tune từ block_10 (BN được cập nhật lại)")
unfreeze_from(model_mbv2, 'block_10')
model_mbv2.compile(optimizer=keras.optimizers.Adam(1e-4),
                   loss='categorical_crossentropy', metrics=['accuracy'])
h_m2 = model_mbv2.fit(DATASETS['mobilenet']['train'],
                      validation_data=DATASETS['mobilenet']['val'],
                      epochs=EPOCHS_FT, callbacks=get_callbacks('mbv2_p2', 7), verbose=1)
TRAIN_TIME['MobileNetV2'] = time.time() - t0
HIST['MobileNetV2'] = merge_hist(h_m2 and h_m1, h_m2)

MODELS = {'Custom CNN' : (model_custom, 'custom'),
          'ResNet50'   : (model_resnet, 'resnet'),
          'MobileNetV2': (model_mbv2,   'mobilenet')}

print("\n" + "="*80); print("✅ HOÀN TẤT HUẤN LUYỆN 3 MODEL"); print("="*80)
for k, v in TRAIN_TIME.items():
    best_val = max(HIST[k]['val_accuracy'])
    print(f"  {k:<14}: {v/60:5.1f} phút | best val_accuracy = {best_val*100:.2f}%")
```

```python
# ====== CELL 13: ĐÁNH GIÁ TRÊN TEST SET ======
RESULTS = {}
y_true = test_labels.copy()

for name, (model, mt) in MODELS.items():
    print(f"\n⏳ Đang dự đoán với {name} ...")
    t0 = time.time()
    probs = model.predict(DATASETS[mt]['test'], verbose=1)
    infer_time = (time.time()-t0)/len(y_true)*1000       # ms/ảnh
    y_pred = probs.argmax(1)
    RESULTS[name] = {
        'probs': probs, 'y_pred': y_pred,
        'acc'  : accuracy_score(y_true, y_pred),
        'f1_macro': f1_score(y_true, y_pred, average='macro'),
        'f1_weighted': f1_score(y_true, y_pred, average='weighted'),
        'infer_ms': infer_time,
    }
    print(f"   ✅ {name}: Test Accuracy = {RESULTS[name]['acc']*100:.2f}% | "
          f"Macro-F1 = {RESULTS[name]['f1_macro']:.4f} | {infer_time:.2f} ms/ảnh")
```

```python
# ====== CELL 15: GRAD-CAM ======
# Lưu ý: hàm prepare_single() đã được định nghĩa ở CELL 8 (dùng pipeline riêng từng model)
LAST_CONV = {
    'Custom CNN' : 'last_conv_act',      # ReLU cuối của Custom CNN
    'ResNet50'   : 'conv5_block3_out',   # block conv cuối ResNet50
    'MobileNetV2': 'out_relu',           # ReLU cuối MobileNetV2
}

def make_gradcam_heatmap(img_array, model, last_conv_name, pred_index=None):
    """Grad-CAM: trọng số hóa feature map cuối bằng gradient trung bình
       của điểm số lớp mục tiêu → bản đồ vùng ảnh model dựa vào để quyết định."""
    grad_model = keras.Model(model.inputs,
                             [model.get_layer(last_conv_name).output, model.output])
    with tf.GradientTape() as tape:
        conv_out, preds = grad_model(img_array, training=False)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]
    grads   = tape.gradient(class_channel, conv_out)
    pooled  = tf.reduce_mean(grads, axis=(0, 1, 2))
    heatmap = tf.squeeze(conv_out[0] @ pooled[..., tf.newaxis])
    heatmap = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy(), int(pred_index), float(preds[0][pred_index])

def overlay(heatmap, disp_img, alpha=0.45):
    hm = cv2.resize(heatmap, (disp_img.shape[1], disp_img.shape[0]))
    hm = cv2.applyColorMap(np.uint8(255*hm), cv2.COLORMAP_JET)
    hm = cv2.cvtColor(hm, cv2.COLOR_BGR2RGB)
    return np.uint8(hm*alpha + disp_img*(1-alpha))

# ----- Kiểm tra tên layer có tồn tại không (tránh lỗi khi TF đổi version) -----
for name, (model, _) in MODELS.items():
    try:
        model.get_layer(LAST_CONV[name])
        print(f"✅ {name:<14} → Grad-CAM layer: '{LAST_CONV[name]}'")
    except ValueError:
        cands = [l.name for l in model.layers
                 if isinstance(l.output_shape, tuple) and len(l.output_shape) == 4][-5:]
        print(f"❌ {name}: không tìm thấy '{LAST_CONV[name]}'. Gợi ý: {cands}")

# ----- Grad-CAM trên cùng một bộ ảnh test cho cả 3 model -----
rs = np.random.RandomState(7)
demo_idx = rs.choice(len(test_paths), 6, replace=False)

for name, (model, mt) in MODELS.items():
    fig, axes = plt.subplots(2, len(demo_idx), figsize=(3.1*len(demo_idx), 7.2))
    for j, idx in enumerate(demo_idx):
        x, disp = prepare_single(test_paths[idx], mt)      # ← pipeline riêng của model
        hm, pi, conf = make_gradcam_heatmap(x, model, LAST_CONV[name])
        true_c = y_true[idx]; ok = (pi == true_c)
        axes[0, j].imshow(disp); axes[0, j].axis('off')
        axes[0, j].set_title(f"True: {CLASS_NAMES[true_c][:18]}", fontsize=9)
        axes[1, j].imshow(overlay(hm, disp)); axes[1, j].axis('off')
        axes[1, j].set_title(f"{'✅' if ok else '❌'} {CLASS_NAMES[pi][:16]}\n{conf*100:.1f}%",
                             fontsize=9, color='green' if ok else 'red')
    plt.suptitle(f"GRAD-CAM — {name}  ({CFG[mt]['desc']})\n"
                 f"hàng trên: ảnh đầu vào sau tiền xử lý | hàng dưới: vùng model tập trung",
                 fontsize=13)
    plt.tight_layout(); plt.show()

# ----- So sánh Grad-CAM của 3 model trên CÙNG 1 ảnh -----
compare_idx = demo_idx[:4]
fig, axes = plt.subplots(4, len(compare_idx), figsize=(3.2*len(compare_idx), 12.5))
for j, idx in enumerate(compare_idx):
    raw = cv2.cvtColor(cv2.imread(test_paths[idx]), cv2.COLOR_BGR2RGB)
    axes[0, j].imshow(raw)
    axes[0, j].set_title(f"GỐC\n{CLASS_NAMES[y_true[idx]][:20]}", fontsize=9)
    for i, (name, (model, mt)) in enumerate(MODELS.items()):
        x, disp = prepare_single(test_paths[idx], mt)
        hm, pi, conf = make_gradcam_heatmap(x, model, LAST_CONV[name])
        ok = (pi == y_true[idx])
        axes[i+1, j].imshow(overlay(hm, disp))
        axes[i+1, j].set_title(f"{name}\n{'✅' if ok else '❌'} {conf*100:.1f}%",
                               fontsize=9, color='green' if ok else 'red')
for a in axes.ravel(): a.axis('off')
plt.suptitle('SO SÁNH VÙNG CHÚ Ý CỦA 3 MÔ HÌNH TRÊN CÙNG MỘT ẢNH', fontsize=15)
plt.tight_layout(); plt.show()
```

```python
# ====== CELL 16: ERROR ANALYSIS ======
def error_analysis(name, top_k=12, n_pairs=10):
    model, mt = MODELS[name]
    y_pred, probs = RESULTS[name]['y_pred'], RESULTS[name]['probs']
    wrong = np.where(y_pred != y_true)[0]
    conf_wrong = probs[wrong, y_pred[wrong]]

    print("\n" + "="*100)
    print(f"ERROR ANALYSIS — {name}")
    print("="*100)
    print(f"Tổng ảnh test: {len(y_true):,} | Dự đoán SAI: {len(wrong):,} "
          f"({len(wrong)/len(y_true)*100:.2f}%)")
    print(f"Độ tự tin trung bình khi SAI: {conf_wrong.mean()*100:.1f}% "
          f"| Số lỗi 'tự tin cao' (>90%): {(conf_wrong>0.9).sum()}")

    # --- Cặp nhầm lẫn phổ biến nhất ---
    cm = confusion_matrix(y_true, y_pred, labels=range(NUM_CLASSES))
    pairs = [(i, j, cm[i, j]) for i in range(NUM_CLASSES) for j in range(NUM_CLASSES)
             if i != j and cm[i, j] > 0]
    pairs.sort(key=lambda t: -t[2])
    print(f"\n🔍 TOP {n_pairs} CẶP NHẦM LẪN NHIỀU NHẤT:")
    for t, p, c in pairs[:n_pairs]:
        print(f"   {c:>4} ảnh | THẬT: [{t:2d}] {CLASS_NAMES[t][:34]:<34} → ĐOÁN: [{p:2d}] {CLASS_NAMES[p][:34]}")

    # --- Class có accuracy thấp nhất ---
    per_cls = cm.diagonal() / cm.sum(1).clip(min=1)
    worst = np.argsort(per_cls)[:8]
    print(f"\n📉 8 CLASS CÓ ĐỘ CHÍNH XÁC THẤP NHẤT:")
    for c in worst:
        print(f"   Class {c:2d} - {CLASS_NAMES[c][:38]:<38} acc = {per_cls[c]*100:5.1f}% "
              f"({cm[c].sum()} ảnh test)")

    # --- Hiển thị ảnh sai + Grad-CAM ---
    sel = wrong[np.argsort(-conf_wrong)][:top_k]      # lỗi "tự tin" nhất → nguy hiểm nhất
    rows = 2; cols = int(np.ceil(top_k/rows))
    fig, axes = plt.subplots(rows*2, cols, figsize=(2.6*cols, 5.6*rows))
    for k, idx in enumerate(sel):
        r, c = divmod(k, cols)
        x_, disp = prepare_single(test_paths[idx], mt)
        hm, pi, conf = make_gradcam_heatmap(x_, model, LAST_CONV[name],
                                            pred_index=int(y_pred[idx]))
        axes[r*2, c].imshow(disp); axes[r*2, c].axis('off')
        axes[r*2, c].set_title(f"THẬT: {CLASS_NAMES[y_true[idx]][:17]}", fontsize=8, color='green')
        axes[r*2+1, c].imshow(overlay(hm, disp)); axes[r*2+1, c].axis('off')
        axes[r*2+1, c].set_title(f"ĐOÁN: {CLASS_NAMES[pi][:17]}\n{conf*100:.1f}%", fontsize=8, color='red')
    for a in axes.ravel(): a.axis('off')
    plt.suptitle(f'{name} — {top_k} ẢNH SAI VỚI ĐỘ TỰ TIN CAO NHẤT (kèm Grad-CAM)', fontsize=14)
    plt.tight_layout(); plt.show()

    # --- Phân tích nguyên nhân theo độ sáng & kích thước ---
    br_w, br_r, sz_w, sz_r = [], [], [], []
    for i in wrong[:600]:
        im = cv2.imread(test_paths[i]);  g = cv2.cvtColor(im, cv2.COLOR_BGR2GRAY)
        br_w.append(g.mean()); sz_w.append(im.shape[0]*im.shape[1])
    right = np.where(y_pred == y_true)[0][:600]
    for i in right:
        im = cv2.imread(test_paths[i]);  g = cv2.cvtColor(im, cv2.COLOR_BGR2GRAY)
        br_r.append(g.mean()); sz_r.append(im.shape[0]*im.shape[1])

    fig, ax = plt.subplots(1, 3, figsize=(18, 4.2))
    ax[0].hist([br_r, br_w], bins=30, label=['Đúng','Sai'], color=['seagreen','crimson'])
    ax[0].set_title('Độ sáng: ảnh đúng vs sai'); ax[0].set_xlabel('Mean pixel'); ax[0].legend()
    ax[1].hist([np.sqrt(sz_r), np.sqrt(sz_w)], bins=30, label=['Đúng','Sai'],
               color=['seagreen','crimson'])
    ax[1].set_title('Kích thước ảnh gốc (√diện tích)'); ax[1].set_xlabel('px'); ax[1].legend()
    ax[2].hist(conf_wrong, bins=30, color='darkorange', edgecolor='k')
    ax[2].set_title('Phân bố độ tự tin khi dự đoán SAI'); ax[2].set_xlabel('Confidence')
    plt.suptitle(f'{name} — NGUYÊN NHÂN LỖI', fontsize=14); plt.tight_layout(); plt.show()

    print(f"\n💡 NHẬN XÉT: ảnh SAI có độ sáng TB = {np.mean(br_w):.1f} (ảnh đúng: {np.mean(br_r):.1f}), "
          f"kích thước TB = {np.sqrt(np.mean(sz_w)):.1f}px (ảnh đúng: {np.sqrt(np.mean(sz_r)):.1f}px)")
    print("   → Lỗi tập trung ở ảnh nhỏ / quá tối / mờ, và ở các nhóm biển báo giống nhau về hình dạng")
    print("     (nhóm speed limit 20-30-50-80-120, nhóm biển tam giác cảnh báo).")

for name in MODELS:
    error_analysis(name)
```
